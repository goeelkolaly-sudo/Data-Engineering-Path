```python
# دالة للتحقق هل الرقم أولي أم لا
def check_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

# قراءة الرقم من المستخدم
num = int(input("Enter an integer: "))

# البحث عن أولي أكبر من الرقم المدخل
next_num = num + 1

while True:
    if check_prime(next_num):
        print("The next prime number is:", next_num)
        break
    next_num = next_num + 1
```

```python
class BanckACCOUNT():
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    def deposit(self, new):
        self.balance += new
        return self.balance


    def withdraw(self, new):
        self.balance -= new
        return self.balance
    

    def show_balance(self):
        print(f"your balance is {self.balance}")    


clint1 = BanckACCOUNT("ypusef", 100)
clint1.deposit(50)
clint1.withdraw(20)
clint1.show_balance()

```

```python
import numpy as np

scores = np.array([[80, 75, 90],
                   [70, 85, 80],
                   [90, 95, 100]])

shape_res = scores.shape
subjects_mean = scores.mean(axis=0)
max_val = scores.max()
min_val = scores.min()
high_grades = scores[scores > 90]
students_avg_gt_80 = scores.mean(axis=1) > 80
scores_plus_grade = scores + 1

print( shape_res)
print( subjects_mean)
print( max_val)
print( min_val)
print( high_grades)
print( students_avg_gt_80)
print( scores_plus_grade)
```

```python
class DataPiple():
    def __init__(self, name, source, destination, records):
        self.n = name
        self.s = source
        self.d = destination
        self.r = records
    def show_info(self):
        print("pipeline :", self.n)    
        print("MySQL :", self.s)    
        print("Data Warehouse :", self.d) 
        print("records :", self.r)

    def add_records(self, new):
            self.n = new
            self.r += self.n
            return self.r

    def states(self, states):
         self.states = states
         if self.states == "running":
            self.states = "completed"  
            return self.states     
                      
pipeline = DataPiple("ggg", "iujkljlk", "iujij", 15000)
pipeline.show_info()          
pipeline.add_records(60)          
pipeline.states("running")          
        
```

```python
class DataEngineer :
    company = "datatech"
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def show_inf(self):
        print("name :", self.name)
        print("salary :", self.salary)
        print("company :", self.company)
    @classmethod
    def change_company(cls,new_company):
        cls.company = new_company  
        return cls.company 
e1 = DataEngineer("yy", 3444)
e2 = DataEngineer("uu", 3444)
e1.change_company("google")    
e1.show_inf()
e2.show_inf()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name 
        self.salary = salary 
    def show_info(self):
        print(self.name)
        print(self.salary)


class DataEngineer(Employee):
        def __init__(self, name, salary, specialization):
            super().__init__( name, salary)

            self.s = specialization


        def show_info(self):
            super().show_info()
            print(self.s)

e1 = DataEngineer("me", 345, "jkhukj")    
e1.show_info()
```
