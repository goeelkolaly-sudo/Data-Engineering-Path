```python
# دالة للتحقق هل الرقم أولي أم لا
def check_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

# قراءة الرقم من المستخدم
num = int(input("Enter an integer: "))

# البحث عن أولي أكبر من الرقم المدخل
next_num = num + 1

while True:
    if check_prime(next_num):
        print("The next prime number is:", next_num)
        break
    next_num = next_num + 1
```

```python
class DataPiple():
    def __init__(self, name, source, destination, records):
        self.n = name
        self.s = source
        self.d = destination
        self.r = records
    def show_info(self):
        print("pipeline :", self.n)    
        print("MySQL :", self.s)    
        print("Data Warehouse :", self.d) 
        print("records :", self.r)

    def add_records(self, new):
            self.n = new
            self.r += self.n
            return self.r

    def states(self, states):
         self.states = states
         if self.states == "running":
            self.states = "completed"  
            return self.states     
                      
pipeline = DataPiple("ggg", "iujkljlk", "iujij", 15000)
pipeline.show_info()          
pipeline.add_records(60)          
pipeline.states("running")          
        
```

```python
class DataSource:
    source_count = 0

    def __init__(self, name, records):
        self.name = name 
        self.__records = records  # متغير محمي (Private)

        DataSource.source_count += 1

    # 5️⃣ Class Method
    @classmethod
    def get_source_count(cls):
        return cls.source_count
    
    # 3️⃣ Encapsulation Getters & Setters
    def get_records(self):
        return self.__records

    def add_records(self, amount):
        if amount > 0: 
            self.__records += amount
        else:
            print("Try again!! Amount must be positive.")    
        return self.__records

    def connect(self):
        print(f"Connecting to {self.name}...")

    def read_data(self):
        print("Reading data from generic data source...")


# 2️⃣ Inheritance & Overriding
class MySQLSource(DataSource):
    def read_data(self):
        print(f"Reading {self.get_records()} records from MySQL")


class CSVSource(DataSource):
    def read_data(self):
        print(f"Reading {self.get_records()} records from CSV")


class APISource(DataSource):
    def read_data(self):
        print(f"Reading {self.get_records()} records from API")


# 6️⃣ & 7️⃣ Pipeline Class
class Pipeline:
    def __init__(self, name, source, destination):
        self.name = name 
        self.source = source 
        self.destination = destination
        self.status = "pending"

    def run(self):
        self.source.connect()    
        self.source.read_data()
        self.status = "completed"

    def show_info(self):
        print(f"Pipeline: {self.name}")
        print(f"Source: {self.source.name}")  
        print(f"Destination: {self.destination}")
        print(f"Records: {self.source.get_records()}")
        print(f"Status: {self.status}")
        
    def is_ready(self):
        return self.source.get_records() >= 5000 and self.status == "completed"

    # 9️⃣ Magic Methods
    def __str__(self):
        return f"Pipeline({self.name}) - Status: {self.status}"

    def __len__(self):
        return self.source.get_records()


# --- 8️⃣ Polymorphism التجربة ---

sources = [
    MySQLSource("MySQL DB", 980),
    CSVSource("Sales File", 477),
    APISource("Users API", 3454)
]

print("--- Polymorphism Loop ---")
for source in sources:
    source.read_data()

print("\n--- Pipeline Testing ---")
# 🔟 إثبات التجربة النهائية
mysql_src = MySQLSource("Main MySQL", 5000)
pipeline1 = Pipeline("Sales Pipeline", mysql_src, "Data Warehouse")

pipeline1.run()
pipeline1.show_info()

print("\n--- Magic Methods Output ---")
print(pipeline1)            # ينادي __str__
print("Len:", len(pipeline1)) # ينادي __len__

print("\nTotal Sources Created:", DataSource.get_source_count())
```
